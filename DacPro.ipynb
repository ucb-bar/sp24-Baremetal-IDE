```python
i2c = I2cController()
i2c.configure('ftdi://ftdi:2232:1:7/1')
i2sdac = i2c.get_port(0x4c)
```

```python
def reset_dac(dac):
    dac.write_to(2, b'\x10')
    dac.write_to(1, b'\x11')
    dac.write_to(2, b'\x00')

def init_dac_44khz(dac):
    dac.write_to(37, b'\x1A')
    dac.write_to(13, b'\x10')
    dac.write_to(14, b'\x10')

    dac.write_to(20, b'\x00') #PLL P
    dac.write_to(21, b'\x20') #PLL J
    dac.write_to(22, b'\x00') #PLL D_MSB
    dac.write_to(23, b'\x00') #PLL D_LSB
    dac.write_to(24, b'\x01') #PLL R
    dac.write_to(27, b'\x01') #DSP CLKDIV
    dac.write_to(28, b'\x0F') #DAC CLKDIV
    dac.write_to(29, b'\x03') #NCP CLKDIV
    dac.write_to(30, b'\x07') #OSR CLKDIV

    dac.write_to(35, b'\x04') #IDAC MSB
    dac.write_to(36, b'\x00') #IDAC LSB

    dac.write_to(34, b'\x00') #FS SPEED MODE

```

```python
reset_dac(i2sdac)
```

```python
init_dac_44khz(i2sdac)
```
